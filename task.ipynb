## Exercise 2: Punctuation
The goal of this exercise is to learn to deal with punctuation.

```python
def remove_punctuation(text):
    return "".join([c for c in text if c not in string.punctuation])

text_with_punct = 'Remove, this from .? the sentence !!!! !"#&\'()*+,-./:;<=>_'
print(remove_punctuation(text_with_punct))
```

## Exercise 3: Tokenization
The goal of this exercise is to learn to tokenize as text.

```python
text = """Bitcoin is a cryptocurrency invented in 2008 by an unknown person or group of people using the name Satoshi Nakamoto. The currency began use in 2009 when its implementation was released as open-source software."""

print("Sentence Tokenization:")
print(sent_tokenize(text))

print("\nWord Tokenization:")
print(word_tokenize(text))
```

## Exercise 4: Stop words
The goal of this exercise is to learn to remove stop words with NLTK.

```python
text = """
The goal of this exercise is to learn to remove stop words with NLTK.  Stop words usually refers to the most common words in a language.
"""
stop_words = set(stopwords.words('english'))
tokens = word_tokenize(text)
filtered_tokens = [w for w in tokens if w.lower() not in stop_words]
print(filtered_tokens)
```

## Exercise 5: Stemming
The goal of this exercise is to learn to use stemming using NLTK.

```python
stemmer = PorterStemmer()
text = """
The interviewer interviews the president in an interview
"""
tokens = word_tokenize(text)
stemmed_tokens = [stemmer.stem(w) for w in tokens]
print(stemmed_tokens)
```

## Exercise 6: Text preprocessing
The goal of this exercise is to learn to create a function to prepocess and clean a text using NLTK.

```python
def preprocess_text(text):
    # 1. Lowercase
    text = text.lower()
    # 2. Removing Punctuation
    text = "".join([c for c in text if c not in string.punctuation])
    # 3. Tokenization
    tokens = word_tokenize(text)
    # 4. Stopword Filtering
    tokens = [w for w in tokens if w not in stop_words]
    # 5. Stemming
    tokens = [stemmer.stem(w) for w in tokens]
    return " ".join(tokens)

text = """
01 Edu System presents an innovative curriculum in software engineering and programming. With a renowned industry-leading reputation, the curriculum has been rigorously designed for learning skills of the digital world and technology industry. Taking a different approach than the classic teaching methods today, learning is facilitated through a collective and co-creative process in a professional environment.
"""
print(preprocess_text(text))
```

## Exercise 7: Bag of Word representation
The goal of this exercise is to understand the creation of a Bag of Word (BoW) model.

```python
# Load data
try:
    with open('tweets_train.txt', 'r') as f:
        tweets = [line.strip() for line in f.readlines() if line.strip()]
except FileNotFoundError:
    print("tweets_train.txt not found. Please provide the file.")
    tweets = []

if tweets:
    # Preprocess
    preprocessed_tweets = [preprocess_text(t) for t in tweets]

    # CountVectorizer
    vectorizer = CountVectorizer(max_features=500)
    X = vectorizer.fit_transform(preprocessed_tweets)

    # Check shape
    print("Shape of word count matrix:", X.shape)

    # Create DataFrame using sparse.from_spmatrix or similar if available, else dense
    # The requirement asks for 'from_spmatrix'
    feature_names = vectorizer.get_feature_names_out()
    try:
        count_vectorized_df = pd.DataFrame.sparse.from_spmatrix(X, columns=feature_names)
    except (AttributeError, ImportError):
        # Fallback if pandas version is old
        count_vectorized_df = pd.DataFrame(X.toarray(), columns=feature_names)

    print("\nFirst 5 rows of DataFrame:")
    print(count_vectorized_df.head())

    # Show token counts of the fourth tweet
    if len(tweets) >= 4:
        print("\nToken counts of the fourth tweet:")
        if hasattr(count_vectorized_df, 'sparse'):
             # Accessing sparse series might be slightly different
             row = count_vectorized_df.iloc[3]
             print(row[row > 0]) # Should handle sparse automatically
        else:
             print(count_vectorized_df.iloc[3][count_vectorized_df.iloc[3] > 0])

    # Show 15 most used words
    sum_words = X.sum(axis=0)
    words_freq = [(word, sum_words[0, idx]) for word, idx in vectorizer.vocabulary_.items()]
    words_freq = sorted(words_freq, key = lambda x: x[1], reverse=True)
    print("\n15 most used words:")
    print(words_freq[:15])

    # Add label column
    # Note: Using placeholder labels as actual labels are not parsed from the file
    count_vectorized_df['label'] = np.random.choice([1, 0, -1], count_vectorized_df.shape[0])
    print("\nDataFrame with labels:")
    print(count_vectorized_df.head())
else:
    print("No data to process.")
```
